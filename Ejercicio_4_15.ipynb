```python
import numpy as np

# 1. Datos del problema
dPhi_dt_dE = 7.5e8  # fotones / (m^2 * s * keV)
t_seg = 3600  # s
Phi_E = dPhi_dt_dE * t_seg  # fotones / (m^2 * keV)

# 2. Datos tabulados del NIST para Aluminio
E_nist_keV = np.array([10, 15, 20, 30, 40, 50, 60, 80, 100])
mu_en_rho_cm2g = np.array(
    [
        2.543e01,
        7.487e00,
        3.094e00,
        8.778e-01,
        3.601e-01,
        1.840e-01,
        1.099e-01,
        5.511e-02,
        3.794e-02,
    ]
)

# Convertir mu_en/rho de cm^2/g a m^2/kg: (1 cm^2/g = 0.1 m^2/kg)
mu_en_rho_m2kg = mu_en_rho_cm2g * 0.1

# Factores de conversión de energía (1 keV = 1.60217663e-16 J)
eV_to_J = 1.60217663e-16

# ==========================================
# Integración Numérica Continua ("Proper")
# ==========================================
E_grid = np.linspace(10, 100, 10000)

log_E_nist = np.log(E_nist_keV)
log_mu_nist = np.log(mu_en_rho_m2kg)
log_E_grid = np.log(E_grid)

mu_en_rho_grid = np.exp(np.interp(log_E_grid, log_E_nist, log_mu_nist))

dK_grid = Phi_E * (E_grid * eV_to_J) * mu_en_rho_grid

# Selección automática según versión de NumPy (trapezoid o trapz)
trapz_func = getattr(np, "trapezoid", getattr(np, "trapz", None))
K_el_exacto = trapz_func(dK_grid, E_grid)

print("-" * 50)
print(f"Kerma de Colisión (Integración exacta continua) = {K_el_exacto:.4f} Gy")
print("-" * 50)
```
